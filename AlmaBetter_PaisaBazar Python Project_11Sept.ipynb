# Age vs Annual Income

```python
unique_customers = df.drop_duplicates(subset=['Customer_ID'])

# Create Age Groups
bins = [10, 20, 30, 40, 50, 60]
labels = ['18-20', '21-30', '31-40', '41-50', '51-60']

unique_customers['Age_Group'] = pd.cut(
    unique_customers['Age'],
    bins=bins,
    labels=labels
)

# Average Annual Income by Age Group
age_income = unique_customers.groupby('Age_Group')['Annual_Income'].mean()

plt.figure(figsize=(8,6))

ax = sns.barplot(
    x=age_income.index,
    y=age_income.values,
    palette='Blues_d'
)

# Add value labels
for i, value in enumerate(age_income.values):
    ax.text(
        i,
        value,
        f'{value:,.0f}',
        ha='center',
        va='bottom'
    )

plt.title("Average Annual Income by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Annual Income")

plt.show()
```

# Credit History Vs Credit Score

```python
unique_customers = df.drop_duplicates(subset=['Customer_ID'])

# Convert to string
unique_customers['Credit_History_Age'] = unique_customers[
    'Credit_History_Age'
].astype(str)

# Extract total months
unique_customers['Credit_History_Months'] = (
    unique_customers['Credit_History_Age']
    .str.extract('(\d+)')[0]
    .astype(float)
)

# Convert months to years
unique_customers['Credit_History_Years'] = (
    unique_customers['Credit_History_Months'] / 12
)

# Average credit history by credit score
credit_history_data = unique_customers.groupby('Credit_Score')[
    'Credit_History_Years'
].mean()

# Print data
print(credit_history_data)

plt.figure(figsize=(8,6))

ax = sns.barplot(
    x=credit_history_data.index,
    y=credit_history_data.values,
    palette='Set2'
)

# Add labels
for i, value in enumerate(credit_history_data.values):
    ax.text(
        i,
        value,
        f'{value:.1f}',
        ha='center',
        va='bottom'
    )

plt.title("Average Credit History by Credit Score")
plt.xlabel("Credit Score")
plt.ylabel("Average Credit History (Years)")

plt.show()
```
